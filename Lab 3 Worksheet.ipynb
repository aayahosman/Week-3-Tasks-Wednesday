```python
from bs4 import BeautifulSoup
import re

with open('bongsoon.html', 'r', encoding='utf-8') as f:
    html_content = f.read()

soup = BeautifulSoup(html_content, 'html.parser')
```

Check that the parsing was successful:

```python
soup.title
```

## Exercise 1

```python
description = soup.find('meta', attrs={'name': 'description'})
text = description['content']

text
```

```python
year_pattern = r'\d{4}'

year_matches = re.findall(year_pattern, text)

year_matches
```

```python
word_pattern = r'\b[A-Za-z]{6}\b'

word_matches = re.findall(word_pattern, text)

word_matches
```

## Exercise 2

```python
synopsis = soup.find('div', class_='show-synopsis')
synopsis_text = synopsis.find('p').get_text(strip=True)

synopsis_text
```

```python
vowel_pattern = r'\b[aeiouAEIOU]\w*\b'

vowel_matches = re.findall(vowel_pattern, synopsis_text)

vowel_matches
```

```python
no_ae_pattern = r'\b[b-df-zB-DF-Z]+\b'

no_ae_matches = re.findall(no_ae_pattern, synopsis_text)

no_ae_matches
```

## Exercise 3

```python
link = soup.find('a', string='Viki')
url = link['href']

url
```

```python
domain_pattern = r'https%3A%2F%2Fwww\.(.+?)%2F'

domain_match = re.search(domain_pattern, url)

domain_match.group(1)
```

```python
url_pattern = r'(https)%3A%2F%2Fwww\.(.+?)%2Ftv%2F(\w+)'

url_match = re.search(url_pattern, url)

url_match.groups()
```

```python
print(url_match.group(1))
print(url_match.group(2))
print(url_match.group(3))
```

## Exercise 5

```python
container = soup.find('div', class_='col-xs-8 col-sm-9 col-lg-9')

metrics_text = container.get_text(" ", strip=True)

metrics_text
```

```python
metrics_pattern = (
    r'Ratings:\s*(?P<rating>\d+\.\d+).*?'
    r'from\s*(?P<rating_users>[\d,]+)\s*users.*?'
    r'Watchers:\s*(?P<watchers>[\d,]+).*?'
    r'Reviews:\s*(?P<review_users>[\d,]+)\s*users'
)

metrics_match = re.search(metrics_pattern, metrics_text)

metrics_match.groupdict()
```
